### For Binary

```python
import pandas as pd

def save_filters_to_mif(file_name):
    # Load the Excel file and skip the first row
    excel_data = pd.read_excel(file_name, header=None, skiprows=1)
    
    # Iterate over each column (filter)
    for i, column in enumerate(excel_data.columns):
        filter_data = excel_data[column]
        
        # Prepare MIF content with only the binary values
        mif_content = ""
        for value in filter_data:
            # Convert values to binary (assume values are 0 or 1)
            binary_value = str(int(value))
            mif_content += f"{binary_value}\n"
        
        # Write to a MIF file
        mif_file_name = f"f_64_{i}.mif"
        with open(mif_file_name, 'w') as mif_file:
            mif_file.write(mif_content)
        print(f"Generated {mif_file_name}")

# Usage
save_filters_to_mif("binary_conv2d_64_filters_filters.xlsx")
```

## For Decimals

```python
import pandas as pd

def save_filters_to_mif(file_name):
    # Load the Excel file and skip the first row
    excel_data = pd.read_excel(file_name, header=None, skiprows=1)
    
    # Iterate over each column (filter)
    for i, column in enumerate(excel_data.columns):
        filter_data = excel_data[column]
        
        # Prepare MIF content with decimal values
        mif_content = ""
        for value in filter_data:
            # Convert the decimal value to a string
            decimal_value = str(value)
            mif_content += f"{decimal_value}\n"
        
        # Write to a MIF file
        mif_file_name = f"f_64_{i}.mif"
        with open(mif_file_name, 'w') as mif_file:
            mif_file.write(mif_content)
        print(f"Generated {mif_file_name}")

# Usage
save_filters_to_mif("processed_filters.xlsx")
```

```python
import pandas as pd

def process_and_save_excel(file_name, output_file):
    # Load the Excel file
    excel_data = pd.read_excel(file_name, header=None)
    
    # Transpose the data (rows become columns, columns become rows)
    transposed_data = excel_data.T
    
    # Remove the first column (which is the first row in original data)
    transposed_data = transposed_data.drop(0, axis=1)
    
    # Add numbers in the first row for each column
    transposed_data.columns = [f"Filter_{i}" for i in range(1, len(transposed_data.columns) + 1)]
    
    # Save the processed data to a new Excel file
    transposed_data.to_excel(output_file, index=False)
    print(f"Processed data saved to {output_file}")

# Usage
process_and_save_excel("conv2d_64_filters_filters.xlsx", "processed_filters.xlsx")
```

## Matrix Formation

```python
import numpy as np

def save_matrix_to_mif(matrix, mif_file_name):
    # Get the number of rows (depth) and columns (width)
    depth, width = matrix.shape

    # Prepare the MIF content with only the binary values
    mif_content = ""
    
    # Loop through each row and convert the values to binary format
    for addr in range(depth):
        # Convert the row to a binary string (assuming each element is a bit or 1 or 0)
        binary_value = ''.join(str(int(x)) for x in matrix[addr])
        mif_content += f"{binary_value}\n"
    
    # Write to the MIF file
    with open(mif_file_name, 'w') as mif_file:
        mif_file.write(mif_content)
    print(f"Generated {mif_file_name}")

# Example: Define a 2x200 matrix with random 0 or 1 values (for illustration)
matrix = np.random.randint(0, 10, size=(200, 2))

# Save the matrix to a MIF file
save_matrix_to_mif(matrix, "matrix_2x200.mif")
```
